```python
# Overall Attrition

attrition_rate = (df["Attrition"].eq("Yes").mean()) * 100
print(f"Attrition Rate: {attrition_rate:.2f}%")
```

```python
# Attrition by Department

dept_attrition = df.groupby("Department").agg(
    Employees=("Attrition", "size"),
    Leavers=("Attrition", lambda x: x.eq("Yes").sum())
)

dept_attrition["Attrition_Rate"] = (
    dept_attrition["Leavers"] / dept_attrition["Employees"] * 100
).round(2)

display(dept_attrition)
```

```python
sns.barplot(data=dept_attrition.reset_index(),
            x="Department", y="Attrition_Rate")

plt.title("Attrition Rate by Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Attrition vs Overtime

display(pd.crosstab(
    df["OverTime"], df["Attrition"], normalize="index"
).mul(100).round(2))
```

```python
# Attrition Rate by Job Role

role_rate = df.groupby("JobRole")["Attrition"].apply(
    lambda x: x.eq("Yes").mean() * 100
).sort_values(ascending=False)

role_rate.plot(kind="bar", title="Attrition Rate by Job Role",
               ylabel="Attrition Rate (%)")
plt.tight_layout()
plt.show()
```

```python
# Compare leavers and Stayers by Age

sns.histplot(data=df, x="Age", hue="Attrition",
             multiple="stack", bins=15)
plt.title("Employee Age Distribution by Attrition")
plt.tight_layout()
plt.show()
```

```python
# Tenure vs Attrition

sns.boxplot(data=df, x="Attrition", y="YearsAtCompany")
plt.title("Years at Company: Leavers vs Stayers")
plt.show()
```

```python
# Monthly Income by Attrition

sns.boxplot(data=df, x="Attrition", y="MonthlyIncome")
plt.title("Monthly Income: Leavers vs Stayers")
plt.show()
```

```python
# Job Satisfaction Scores by Attrition

sns.countplot(data=df, x="JobSatisfaction", hue="Attrition")
plt.title("Job Satisfaction by Attrition")
plt.show()
```
